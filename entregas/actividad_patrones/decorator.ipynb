#INCORRECTO
queremos añadir funcionalidades adicionales a un mensaje(como encriptarlo o verlo en mayuscula), pero aqui se busca resolver el problema a la fuera con herencia

```python

class Mensaje:
    def enviar(self, texto):
        print(f'Enviando: {texto}')

class MensajeEncriptado(Mensaje):
    def enviar(self, texto):
        texto = f'***{texto}***'
        super().enviar(texto)

class MensajeMayusculas(Mensaje):
    def enviar(self, texto):
        texto = texto.upper()
        super().enviar(texto)

class MensajeEncriptadoYMayusculas(Mensaje):
    def enviar(self, texto):
        texto = f'***{texto.upper()}***'
        super().enviar(texto)
```

```python
# PRUEBA
msg = MensajeEncriptadoYMayusculas()
msg.enviar('Hola')
```

#CORRECTO
decorador permite envolver el envio del mensaje agregandole encriptacion, mayusculas o lo que se necesite

@startuml

abstract class Mensaje {
    + enviar(texto: str): None
}

class MensajeBase {
    + enviar(texto: str): None
}

class DecoradorMensaje {
    - _mensaje: Mensaje
    + enviar(texto: str): None
}

class Encriptado {
    + enviar(texto: str): None
}

class Mayusculas {
    + enviar(texto: str): None
}

Mensaje <|-- MensajeBase
Mensaje <|-- DecoradorMensaje

DecoradorMensaje <|-- Encriptado
DecoradorMensaje <|-- Mayusculas

DecoradorMensaje o-- Mensaje

@enduml

```python
from abc import ABC, abstractmethod

class Mensaje(ABC):
    @abstractmethod
    def enviar(self, texto: str) -> None: ...

class MensajeBase(Mensaje):
    def enviar(self, texto: str) -> None:
        print(f'Enviando: {texto}')

class DecoradorMensaje(Mensaje):
    def __init__(self, mensaje: Mensaje):
        self._mensaje = mensaje

    def enviar(self, texto: str) -> None:
        self._mensaje.enviar(texto)

class Encriptado(DecoradorMensaje):
    def enviar(self, texto: str) -> None:
        texto = f'***{texto}***'
        self._mensaje.enviar(texto)

class Mayusculas(DecoradorMensaje):
    def enviar(self, texto: str) -> None:
        texto = texto.upper()
        self._mensaje.enviar(texto)
```

```python

mensaje = Mayusculas(Encriptado(MensajeBase()))
mensaje.enviar("hola")
```
