#INCORRECTO

```python

class Notificador:
    def __init__(self, medio):
        self.medio = medio

    def enviar(self, mensaje):
        if self.medio == "email":
            print(f"[EMAIL]: {mensaje}")
        elif self.medio == "sms":
            print(f"[SMS]: {mensaje}")
        else:
            print(f"Medio no soportado")
```

```python
# PRUEBA
notificador = Notificador("email")
notificador.enviar("Alerta de seguridad")

notificador.medio = "sms"
notificador.enviar("Alerta de seguridad")
```

#CORRECTO


@startuml

abstract class EstrategiaNotificacion {
    + enviar_mensaje(mensaje: str): None
}

class NotificacionEmail {
    + enviar_mensaje(mensaje: str): None
}

class NotificacionSMS {
    + enviar_mensaje(mensaje: str): None
}

class Notificador {
    + estrategia: EstrategiaNotificacion
    + enviar(mensaje: str): None
}

EstrategiaNotificacion <|-- NotificacionEmail
EstrategiaNotificacion <|-- NotificacionSMS

Notificador o-- EstrategiaNotificacion

@enduml

```python
from abc import ABC, abstractmethod

class EstrategiaNotificacion(ABC):
    @abstractmethod
    def enviar_mensaje(self, mensaje: str) -> None: ...

class NotificacionEmail(EstrategiaNotificacion):
    def enviar_mensaje(self, mensaje: str) -> None:
        print(f"[EMAIL]: {mensaje}")

class NotificacionSMS(EstrategiaNotificacion):
    def enviar_mensaje(self, mensaje: str) -> None:
        print(f"[SMS]: {mensaje}")

class Notificador:
    def __init__(self, estrategia: EstrategiaNotificacion):
        self.estrategia = estrategia

    def enviar(self, mensaje: str) -> None:
        self.estrategia.enviar_mensaje(mensaje)
```

```python
notificador = Notificador(NotificacionEmail())
notificador.enviar("Hola, email")
notificador.estrategia = NotificacionSMS()
notificador.enviar("Hola, SMS")
```
