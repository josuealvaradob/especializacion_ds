```python

class ServicioPagos:
    def enviar_mensaje(self, tipo: str, formadepago: str):
        if tipo == "paypal":
            print(f"[PAYPAL] la forma de pago es: {formadepago}")
        elif tipo == "bbva":
            print(f"[BBVA] la forma de pago es: {formadepago}")
        else:
            print(f"Error: forma de pago no esta inscrita '{tipo}' ")
```

#CORRECTO
con el metodo se delega la creacion de objetos a las subclases mediante un metodo abstracto, y asi se incorporan los nuevos canales sin tocar la ogica del codigo del cliente

@startuml

abstract class MetodoPago {
    + pagar(monto: float): str
}

abstract class CreadorMetodoPago {
    + crear(): MetodoPago
    + procesar(monto: float): str
}

class PagoTarjetaCredito {
    + pagar(monto: float): str
}

class PagoPayPal {
    + pagar(monto: float): str
}

class PagoTransferencia {
    + pagar(monto: float): str
}

class CreadorTarjetaCredito {
    + crear(): MetodoPago
}

class CreadorPayPal {
    + crear(): MetodoPago
}

class CreadorTransferencia {
    + crear(): MetodoPago
}

MetodoPago <|-- PagoTarjetaCredito
MetodoPago <|-- PagoPayPal
MetodoPago <|-- PagoTransferencia

CreadorMetodoPago <|-- CreadorTarjetaCredito
CreadorMetodoPago <|-- CreadorPayPal
CreadorMetodoPago <|-- CreadorTransferencia

CreadorTarjetaCredito ..> PagoTarjetaCredito
CreadorPayPal ..> PagoPayPal
CreadorTransferencia ..> PagoTransferencia

@enduml

```python
from abc import ABC, abstractmethod

class MetodoPago(ABC):
    @abstractmethod
    def pagar(self, monto: float) -> str: ...

class CreadorMetodoPago(ABC):
    @abstractmethod
    def crear(self) -> MetodoPago: ...

    def procesar(self, monto: float) -> str:
        metodo = self.crear()
        return metodo.pagar(monto)

class PagoTarjetaCredito(MetodoPago):
    def pagar(self, monto: float) -> str:
        return f"[TARJETA DE CRÉDITO]: Pago de ${monto:.2f} procesado"

class PagoPayPal(MetodoPago):
    def pagar(self, monto: float) -> str:
        return f"[PAYPAL]: Pago de ${monto:.2f} procesado"

class PagoTransferencia(MetodoPago):
    def pagar(self, monto: float) -> str:
        return f"[TRANSFERENCIA]: Pago de ${monto:.2f} procesado"

class CreadorTarjetaCredito(CreadorMetodoPago):
    def crear(self) -> MetodoPago:
        return PagoTarjetaCredito()

class CreadorPayPal(CreadorMetodoPago):
    def crear(self) -> MetodoPago:
        return PagoPayPal()

class CreadorTransferencia(CreadorMetodoPago):
    def crear(self) -> MetodoPago:
        return PagoTransferencia()
```

```python

creador = CreadorPayPal()
print(creador.procesar(150.0))
```
