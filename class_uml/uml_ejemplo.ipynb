## Explicación detallada

Los diagramas de clases y de secuencia son complementarios: representan dos vistas distintas del mismo sistema.

- **Diagrama de clases (vista estática)**: muestra la *estructura* del sistema — qué clases existen, qué atributos y métodos tiene cada una, y qué tipo de relación (dependencia, asociación, agregación o composición) las conecta. Es como el plano de un edificio: describe las piezas y cómo encajan, pero no dice en qué orden se usan.

- **Diagrama de secuencia (vista dinámica)**: muestra el *comportamiento* del sistema — el orden temporal en que los objetos se envían mensajes para completar un caso de uso concreto. Es como el guion de una escena: describe qué pasa, en qué orden, pero no cómo están construidas las piezas.

### Cómo se conectan ambas vistas en nuestro dominio

En `ejemplo_uml.py` tenemos tres clases: `Course`, `Student` y `Professor`.

1. **Vista estática (`uml_class_diagram.puml`)**: `Professor` tiene una **asociación** con `Student` (lo guarda en `self.student`) y una **dependencia** con `Course` (lo recibe solo como parámetro de `teach()`). Esta estructura es la que hace *posible* la interacción.

2. **Vista dinámica (`uml_sequence_diagram.puml`)**: cuando se ejecuta `professor.teach(course)`, esa estructura se activa como una secuencia concreta de mensajes: primero `Professor` le pide el conocimiento a `Course` (`get_knowledge()`), y luego se lo transmite a `Student` (`remember(knowledge)`).

En otras palabras: el diagrama de clases define *qué relaciones son posibles*, y el diagrama de secuencia muestra *un ejemplo concreto de cómo se usan esas relaciones* en tiempo de ejecución. Documentar un sistema con ambos diagramas da una visión completa: estructura y comportamiento.

## Ejemplo práctico

El siguiente código define las clases `Course`, `Student` y `Professor`, ejecuta el flujo `professor.teach(course)`, y genera ambos diagramas (de clases y de secuencia) con Graphviz para esa misma estructura y flujo.

```python
# Ejemplo de clases y visualización de diagramas UML en Python

from graphviz import Digraph

# Definición de clases
class Course:
    def get_knowledge(self) -> str:
        return "knowledge"

class Student:
    def remember(self, knowledge: str) -> None:
        print(f"El estudiante ha recordado {knowledge}")

class Professor:
    def __init__(self, student: Student) -> None:
        self.student = student
    def teach(self, course: Course) -> None:
        self.student.remember(course.get_knowledge())

# Instanciación y uso
student = Student()
course = Course()
professor = Professor(student)
professor.teach(course)

# Diagrama de clases UML con graphviz
class_dot = Digraph(comment='Diagrama de Clases')
class_dot.node('Course', 'Course')
class_dot.node('Student', 'Student')
class_dot.node('Professor', 'Professor')
class_dot.edge('Professor', 'Student', label='asociación')
class_dot.edge('Professor', 'Course', label='usa', style='dashed')
class_dot

# Diagrama de secuencia UML con graphviz
seq_dot = Digraph('sequence')
seq_dot.attr(rankdir='LR')
for part in ['Professor', 'Student', 'Course']:
    seq_dot.node(part, part)
seq_dot.edge('Professor', 'Course', 'get_knowledge()')
seq_dot.edge('Professor', 'Student', 'remember(knowledge)')
seq_dot
```
